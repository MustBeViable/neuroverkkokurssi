```python
import numpy as np

embeddings = {}

with open("./glove/glove.6B.100d.txt", encoding="utf-8") as f:
    for line in f:
        parts = line.split()
        word = parts[0]
        vector = np.asarray(parts[1:], dtype=np.float32)
        embeddings[word] = vector

print(embeddings["man"].shape)
print(embeddings["woman"].shape)
print(embeddings["king"].shape)
```

```python
vec_man = embeddings["man"]
vec_woman = embeddings["woman"]
vec_king = embeddings["king"]
```

```python
calculus = vec_woman - vec_man + vec_king

print(calculus)
```

```python
import numpy as np

def nearest_words(target_vector, embeddings, top_n, exclude=()):
    similarities = []

    for word, vector in embeddings.items():
        if word in exclude:
            continue

        # Kosinisamankaltaisuus
        denominator = np.linalg.norm(target_vector) * np.linalg.norm(vector)

        if denominator == 0:
            continue

        similarity = np.dot(target_vector, vector) / denominator
        similarities.append((word, similarity))

    # Suurimman samankaltaisuuden sanat ensin
    similarities.sort(key=lambda x: x[1], reverse=True)

    return similarities[:top_n]
```

```python
nearest_words(calculus, embeddings, 10, ("king",))
```

## woman - man + king

GloVe-mallissa sanat on esitetty numeerisina vektoreina, jotka kuvaavat niiden merkityksiä ja yhteyksiä muihin sanoihin. Jos woman-vektorista vähennetään man-vektori, saadaan niiden vektorien välinen ero. Lisättäessä king-vektori saadaan uusi vektori, joka muistuttaa kuninkaallista naista -> queen ja muita vastaavia sanoja (princess, elizabeth). Tuloksissa löytyy myös muita naisiin liittyviä sanoja, kuten daughter ja mother, sekä kuninkaallisuuteen liittyviä sanoja, kuten throne ja prince. Näistä voidaan päätellä, että lopullisessa vektorissa näkyy sekä sukupuoleen että kuninkaallisuuteen liittyviä merkityksiä. Queen löytyi lähimmäksi, ja sen kosinisamankaltaisuus oli n. 0,78.

Oletetaan tämän perusteella, että sama toimii myös sukulaisuussuhteissa. Etsitään sana "brother", vähennetään siitä "man" ja lisätään "woman" -> pitäisi tulla tällä logiikalla sister.

```python
vec_brother = 0

for key in embeddings.keys():
    if key == "brother":
        print("löytyi")
        vec_brother = embeddings[key]
```

```python
assumably_sister = vec_brother - vec_man + vec_woman

nearest_words(assumably_sister, embeddings, 10, ("brother",))
```

Samankaltaisuuksissa sister on hyvin vahva (n. 0,83), mutta ylempää löytyy myös loogiset daughter, wife ja mother. Husband ja father ovat mielenkiintoisia, kun ne tulevat ennen sister, vaikka laskutoimituksella yritetään saada naispuolista vastinetta sanalle brother. Tämä kertoo siitä, ettei vektorien vähentäminen ja lisääminen aina tuota juuri haluttua sanaa, vaan tuloksissa näkyy myös muita sukulaisuuteen liittyviä sanoja.

Etsitään seuraavaksi "he" ja "she" ja kokeillaan samaa, jos sister löytyisi ylempää.

```python
vec_he = 0
vec_she = 0

for key in embeddings.keys():
    if key == "he":
        print("he löytyi")
        vec_he = embeddings[key]
    if key == "she":
        print("she löytyi")
        vec_she = embeddings[key]
```

```python
sister2 = vec_brother - vec_he + vec_she

nearest_words(sister2, embeddings, 10, ("brother",))
```

Tässä "he"- ja "she"-vektorien käyttö näyttää toimivan paremmin halutun sanan löytämisessä kuin "man" ja "woman". Listan alusta löytyy daughter, mother, sister ja wife ennen maskuliinisia sanoja. Sisterin samankaltaisuus oli nyt parempi (n. 0,87 verrattuna edelliseen n. 0,83), ja se nousi sijalta 6 sijalle 3, vaikka daughter ja mother löytyvät edelleen ylempää. Tämän perusteella "he" ja "she" näyttävät kuvaavan tässä esimerkissä haluttua sukupuolten välistä eroa paremmin.

Kokeillaan muita yhtäläisyyksiä, esim. pääkaupunkien löytäminen pariisi - ranska + suomi.

```python
vec_suomi = 0
vec_ranska = 0
vec_pariisi = 0

for key in embeddings.keys():
    if key == "finland":
        print("suomi löytyi")
        vec_suomi = embeddings[key]
    if key == "france":
        print("ranska löytyi")
        vec_ranska = embeddings[key]
    if key == "paris":
        print("pariisi löytyi")
        vec_pariisi = embeddings[key]
```

```python
helsinki = vec_pariisi - vec_ranska + vec_suomi

nearest_words(helsinki, embeddings, 10, ("paris",))
```

Suomen pääkaupunki tuli ykköseksi (n. 0,83), joten tässä tapauksessa pääkaupungista vähennettäessä sen maa ja siihen lisättäessä toinen maa, löytyy lisätyn maan pääkaupunki. Tuloksissa löytyy myös muita pääkaupunkeja, kuten Stockholm ja Oslo. Tämä viittaa siihen, että sanavektorit pystyvät kuvaamaan myös maiden ja pääkaupunkien välisiä suhteita.

Kokeillaan samaa verbien muotoihin. Eli etsitään verbit swim ja swimming sekä walk ja drive. Vähennetään swimming-vektorista swim-vektori, jolloin saadaan niiden sanamuotojen välistä eroa kuvaava vektori, joka voi kuvata myös verbin "ing" muotoon muuttumista. Siihen lisätään haluttu verbi, jolloin pitäisi saada sitä vastaava "ing" muoto, esim. walking tai driving.

```python
vec_uinti = 0
vec_ajo = 0
vec_kävely = 0
vec_uiminen = 0

for key in embeddings.keys():
    if key == "swim":
        print("Uinti löytyi")
        vec_uinti = embeddings[key]
    if key == "swimming":
        print("Uiminen löytyi")
        vec_uiminen = embeddings[key]
    if key == "drive":
        print("Ajo löytyi")
        vec_ajo = embeddings[key]
    if key == "walk":
        print("Kävely löytyi")
        vec_kävely = embeddings[key]
```

```python
print("-" * 20, "Käveleminen", "-" * 20, "\n")

käveleminen = nearest_words(vec_uiminen - vec_uinti + vec_kävely, embeddings, 10, ("swimming", "swim", "walk"))

for i in käveleminen:
    print(i)
print("")

print("-" * 20, "Ajaminen", "-" * 20, "\n")

ajaminen = nearest_words(vec_uiminen - vec_uinti + vec_ajo, embeddings, 10, ("swimming", "swim", "drive"))

for i in ajaminen:
    print(i)
```

Käveleminen eli "walking" sai suuremman kosinisamankaltaisuuden kuin ajaminen "driving" (n. 0,79 verrattuna n. 0,67). Molemmat löytyivät kuitenkin listan ensimmäiseltä sijalta, joten -ing-muodon löytäminen onnistui molemmissa tapauksissa. Lisäksi listojen yläpäästä löytyy walks ja drives, jotka ovat myös samojen verbien muita taivutusmuotoja. Tämä kertoo siitä, että sanavektorit pystyvät kuvaamaan myös verbien eri muotojen välisiä yhteyksiä suhteellisen hyvin.

Lisäksi löytyy vaikeammin selitettäviä sanoja, esim. standing ja setting. Ne ovat myös -ing-muotoja, joista standing liittyy kävelemiseen, mutta tarkoittaa eri asiaa. Setting löytyy molemmista tuloksista, vaikka sen merkityksellinen yhteys laskutoimituksiin ei ole yhtä selkeä. Tämä voi johtua siitä, että sanavektoreissa näkyvät myös sanojen käyttöyhteydet, eikä pelkästään kieliopillinen muoto. Kaikki lähimmät sanat eivät siis välttämättä vastaa suoraan haluttua merkitystä.
